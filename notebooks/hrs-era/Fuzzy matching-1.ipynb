```python
from fuzzywuzzy import fuzz
from fuzzywuzzy import process
import pandas as pd

matching_list_1 = pd.read_excel("C:\\Users\\USER\\Documents\\misc\\2900_green_hotels_cert.xlsx" ,
                                sheet_name = 'clean',
                                header=0)
matching_list_1.head()
```

```python
matching_list_2 = matching_list_1
matching_list_2.head()
```

```python
list_string = matching_list_1['CSR_CERTIFIED_BY'].values.tolist()
len(list_string)
```

```python
choices = matching_list_2['CSR_CERTIFIED_BY'].values.tolist()
len(choices)
```

```python
import time
possibilities = []
start_time = time.time()

for string in list_string:
    #print(string)
    possibility = process.extract(string, choices, limit=3, scorer=fuzz.token_sort_ratio)
    possibilities.append(possibility)
end_time = time.time()
print(end_time - start_time)
```

```python
temp_df = pd.DataFrame(possibilities)
temp_df['Actual_string'] = list_string
temp_df.columns = ['Option-1', 'Option-2','Option-3', 'Actual_string']
temp_df.head()
```

```python
temp_df.to_excel('C:\\Users\\USER\\Documents\\misc\\Env_cert_list.xlsx', 
                  sheet_name = 'Match',
                  header = True,
                  encoding='utf-8',
                  index=False)
```
