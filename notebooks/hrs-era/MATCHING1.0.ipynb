```python
import pyexasol
import configparser

config = configparser.ConfigParser()
config.read('C:\\Users\\USER\\.spyder-py3\\ExasolConnect.ini')
dsn=config['exasolIUT']['dsn']
user=config['exasolIUT']['user']
pwd=config['exasolIUT']['pwd']
schema=config['exasolIUT']['schema']

connect = pyexasol.connect(dsn=dsn, user=user, password=pwd, schema=schema)
print("Connection Success")
```

```python
from fuzzywuzzy import fuzz
from fuzzywuzzy import process
import pandas as pd

QUERY_1 = connect.execute("""SELECT CITY_ID,
                                   CITY_NAME,
                                   COUNTRY_NAME
                                   FROM DWHBIL.LKP_CITY WHERE CITY_ID NOT IN (
                            select 
                                   C.CITY_ID
                            --       C.CITY_NAME,
                            --       H.HOTEL_CITY_ID,
                            --       H.HOTEL_CITY_NAME_DFLT
                                   from DWHBIL.LKP_CITY C
                            INNER JOIN DWHBIL.LKP_HOTEL_CITY H ON H.HOTEL_CITY_NAME_DFLT = C.CITY_NAME)""")

# Importing data into a DataFrame
query1_df = pd.DataFrame()
a=[]
for row in QUERY_1:
    a.append(row)
#print(len(a))
query1_df = pd.DataFrame(a)
df_col_names = QUERY_1.col_names
query1_df.columns = df_col_names
print(query1_df.shape)
```
