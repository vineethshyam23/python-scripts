```python
import ibm_db
import ibm_db_dbi
import configparser
import pandas as pd

config = configparser.ConfigParser()
config.read('C:\\Users\\svi02\\.spyder-py3\\DB2STAT_connect.ini')

conn_str=config['DB2STAT']['conn_str']
ibm_db_conn = ibm_db.connect(conn_str,'','')
```

```python
# Fetch data using ibm_db_dbi
QUERY="""SELECT L.H_KEY                      AS HOTEL_ID,
       DW.H_LOCK_STATUS             AS HOTEL_LOCK_STATUS_ID,
       DW.H_LOCK_STATUS_DESC        AS H_LOCK_STATUS_DESC,
       D.HJ_STATUS                  AS HOTEL_STATUS_ID,
       D.HJ_STATUS_DESCRIPTION_EN   AS HOTEL_STATUS_DESC,
       D.BOOKABLE                   AS IS_BOOKABLE
FROM HRSDB.HOTEL_LOCK_STATUS_DA L
INNER JOIN HRSDB.HOTEL_LOCK_STATUS_DESCRIPTION_LU DW ON L.H_LOCK_STATUS=DW.H_LOCK_STATUS
INNER JOIN DWH.HOTEL_WITH_DEFAULT_LOCATION D ON D.H_KEY=L.H_KEY
WHERE D.BOOKABLE = 0 AND D.HJ_STATUS <> 0 AND DW.H_LOCK_STATUS = 111"""

conn = ibm_db_dbi.Connection(ibm_db_conn)
df = pd.read_sql(QUERY, conn)
print(df.head())
```
