```python
import pyexasol
import configparser

#Location of the ini file
config = configparser.ConfigParser()
config.read('C:\\Users\\USER\\.spyder-py3\\ExasolIUT.ini')
```

```python
dsn=config['exasolIUT']['dsn']
user=config['exasolIUT']['user']
pwd=config['exasolIUT']['pwd']
schema=config['exasolIUT']['schema']
print("SUCCESS")
```

```python
# Exasol connection
connect = pyexasol.connect(dsn=dsn, user=user, password=pwd, schema=schema)
```

```python
sql_query =  input("Please enter the query: ")
```

```python
import_script = 'C:\\Users\\USER\\Documents\\SQL Scripts\\hotel_content_report_exa.sql'

script_open = open(import_script, 'r')
script_read = script_open.read()
#file_read
QUERY = connect.execute(script_read)
```

```python
# Importing data into a DataFrame
import pandas as pd
df = pd.DataFrame()
a=[]
for row in QUERY:
    a.append(row)
#print(len(a))
df = pd.DataFrame(a)
df_col_names = QUERY.col_names
df.columns = df_col_names
#print(df.head())
```

```python
df.to_csv('C:\\Users\\USER\\Downloads\\Hotel_content_report_12062019.csv', 
          sep = ',', 
          header = True,
          encoding='utf-8',
          index=False)
```
