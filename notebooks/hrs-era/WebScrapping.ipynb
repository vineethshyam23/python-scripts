```python
import requests
from bs4 import BeautifulSoup
url = "https://blog.hotelscombined.com/travelling-green-the-worlds-top-eco-friendly-hotels"
page = requests.get(url)
```

```python
print(page.content)
```

```python
soup = BeautifulSoup(page.content, 'html.parser')
print(soup.prettify())
```

```python
hotels_combined = []
for link in soup.find_all('figcaption', class_='wp-caption-text'):
      hotels_combined.append(link.text)
        
hotels_combined[0:5]
```
