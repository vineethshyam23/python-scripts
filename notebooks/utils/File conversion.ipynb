```python
import os
import glob
import pandas as pd

def convert_file(path):
    path = path
    extension = 'csv'
    os.chdir(path)
    result = glob.glob('*.{}'.format(extension))
    split_result = [i.split('.', 1)[0] for i in result]
    for result in split_result:
        df = pd.read_csv(path+result+".csv" ,
                     sep="\t",
                     encoding = 'unicode_escape')
        df.to_csv(path+result+"_1.csv", 
              sep = ',', 
              header = True,
              encoding='utf-8',
              quoting=1,
              index=False)
        print(result)
```

```python
convert_file('T:/Ashwini/')
```
