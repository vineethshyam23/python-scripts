```python
def solution(a):
    # write your code in Python 3.6
    z = []
    for i in range(len(a)):
        for j in range(len(a)):
            if a[i]==a[j] and i!=j:
                    c = [a[i],a[j]]
                    z.append(c)
    return int(len(z)/2)
```
