```python
import interpret
import pandas as pd
import sklearn
from interpret import show
from interpret.glassbox import RegressionTree
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
```

```python
df = pd.read_csv("C:/Users/USER/Documents/Python Scripts/diabetes.csv")
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['Pregnancies', 'Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age']], df['Outcome'], test_size=0.3)
#Creating Model
reg_tree = RegressionTree(random_state=42)
reg_tree.fit(X_train, y_train)
```

```python
#global interpretation
reg_tree_global = reg_tree.explain_global(name='Regression Tree')
show(reg_tree_global)
```

```python
from interpret import perf
reg_tree_perf = perf.RegressionPerf(reg_tree.predict).explain_perf(X_test, y_test, name='Regression Tree')
reg_tree_local = reg_tree.explain_local(X_train, y_train, name='Regression Tree')
reg_tree_all = reg_tree.available_explanations
show(reg_tree_perf)
```

```python
from interpret.glassbox import LinearRegression
lin_reg = LinearRegression(random_state=42)
lin_reg.fit(X_train, y_train)
lin_reg_perf = perf.RegressionPerf(lin_reg.predict).explain_perf(X_test, y_test, name='Linear Regression')
lin_reg_global = lin_reg.explain_global(name='Linear Regression')
lin_reg_local = lin_reg.explain_local(X_train, y_train, name='Linear Regression')
lin_reg_all = lin_reg.available_explanations
```

```python
show([reg_tree_global, reg_tree_local, reg_tree_perf, lin_reg_local, lin_reg_global, lin_reg_perf])
```
