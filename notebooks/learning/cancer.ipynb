```python
import pandas as pd
import numpy as np
#from multiscorer import MultiScorer

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.model_selection import cross_val_score
from numpy import average

#scorer = MultiScorer({                                               # Create a MultiScorer instance
#  'accuracy': (accuracy_score, {}),
#  'precision': (precision_score, {'average': 'weighted'}),
#     'recall': (recall_score, {'average': 'weighted'}),
#       'f1': (f1_score, {'average': 'weighted'})
#})

train_data = pd.read_csv("C:\\Users\\USER\\Downloads\\candidate_data\\train_data.csv" ,
                                header=0)
train_data.describe()
```

```python
test_data = pd.read_csv("C:\\Users\\USER\\Downloads\\candidate_data\\test_data.csv" ,
                                header=0)
```

```python
test_data['cancer_type']= pd.read_csv("C:\\Users\\USER\\Downloads\\candidate_data\\sample_submission.csv" ,
                                header=0)
train_data.head()
```

```python
from datetime import datetime, timedelta
# Date type
train_data['treatment_date'] = train_data['treatment_date'].astype('datetime64[ns]')
train_data['diagnose_date'] = train_data['diagnose_date'].astype('datetime64[ns]')
test_data['treatment_date'] = test_data['treatment_date'].astype('datetime64[ns]')
test_data['diagnose_date'] = test_data['diagnose_date'].astype('datetime64[ns]')

# Interval column
train_data['diagonse_interval'] = train_data['treatment_date'] - train_data['diagnose_date']
test_data['diagonse_interval'] = test_data['treatment_date'] - test_data['diagnose_date']

# String
train_data['diagonse_interval'] = train_data['diagonse_interval'].dt.days.astype(int)
test_data['diagonse_interval'] = test_data['diagonse_interval'].dt.days.astype(int)
# train_data['texture_2'] = train_data['texture_2'].astype(float)
# test_data['texture_2'] = test_data['texture_2'].astype(float)
```

```python
# Using Skicit-learn to split data into training and testing sets
from sklearn.model_selection import train_test_split
# Split the data into training and testing sets
train_features = train_data[train_data.columns[~train_data.columns.isin(['cancer_type', 'diagnose_date', 'treatment_date'])]] 
test_features = test_data[test_data.columns[~test_data.columns.isin(['cancer_type', 'diagnose_date', 'treatment_date'])]] 
train_labels = train_data.iloc[:,-1]
test_labels = test_data.iloc[:,-1]
train_features[train_features.texture_2 =='xx'] = train_features[train_features.texture_2 =='xx'].replace('xx',0)
test_features[test_features.texture_2 =='xx'] = test_features[test_features.texture_2 =='xx'].replace('xx',0)
train_features=train_features.fillna(0)
test_features=test_features.fillna(0)
```

```python
# Import the model we are using
from sklearn.ensemble import RandomForestRegressor
# Instantiate model with 1000 decision trees
rf = RandomForestRegressor(n_estimators = 1000, random_state = 42)
# Train the model on training data
rf.fit(train_features, train_labels)
```

```python
from sklearn.metrics import classification_report
predicted = rf.predict(test_features)
report = classification_report(test_labels, predicted)
```
